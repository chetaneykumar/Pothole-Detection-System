```python
import tensorflow as tf
```

```python
# Load images
data = tf.keras.utils.image_dataset_from_directory(
    r"C:\Chetanaya\Project\Normal pothole dataset\Pothole and normal roads dataset\Train",
    image_size=(64, 64),
    batch_size=2
)
```

```python
# CNN model
model = tf.keras.Sequential([
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Conv2D(16, (3,3), activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(2, activation="softmax")
])
# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Train
model.fit(data, epochs=10)
model.summary()
```

```python
# Prediction
img = tf.keras.utils.load_img("Test/NR121.jpg", 
                              target_size=(64,64))
x = tf.keras.utils.img_to_array(img)
x = tf.expand_dims(x, 0)

p = model.predict(x)
print(data.class_names[p.argmax()])
```

```python
model.save('Pothole_Detector.keras')

import json

# Save class names to a JSON file
with open("classes.json", "w") as f:
    json.dump(data.class_names, f)
```
